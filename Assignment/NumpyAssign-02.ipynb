Q1.
A class has 4 students and 3 subjects. Their marks are given below. The examiner announces a subject-wise bonus of 5, 2 and 8 marks respectively. Add the bonus to every student in one line using broadcasting, then cap any mark above 100 back to 100 and report how many marks had to be capped.

marks = np.array([[88, 96, 95],
                  [72, 99, 91],
                  [98, 85, 97],
                  [60, 75, 80]])

```python
# Q1

marks = np.array([[88, 96, 95],
                  [72, 99, 91],
                  [98, 85, 97],
                  [60, 75, 80]])

bonus = np.array([5, 2, 8])

marks = marks + bonus

marks = np.minimum(marks, 100)

print(marks)

capped = np.sum(marks == 100)

print("Number of marks capped:", capped)
```

Q4.
Two arrays of size 4 are given. First join them with hstack and print the shape. Then, using the same two arrays, build a proper (4, 2) two-column table. In a comment, state why the first attempt did not produce two columns.

heights = np.array([172, 158, 181, 166])
weights = np.array([68, 52, 84, 59])

```python
# Q4

heights = np.array([172, 158, 181, 166])
weights = np.array([68, 52, 84, 59])

first = np.hstack((heights, weights))

print("First attempt:")
print(first)
print("Shape:", first.shape)

table = np.column_stack((heights, weights))

print("\nProper 2-column table:")
print(table)
print("Shape:", table.shape)

# np.hstack joins two 1-D arrays into one longer 1-D array,
# so its shape is (8,), not (4, 2).
```

Q5.
Given a 1D array of 12 monthly sales figures, reshape it into 4 quarters of 3 months each and find which quarter (index) had the highest average sales, along with that average.

sales = np.array([120, 135, 98, 160, 175, 190, 88, 92, 105, 210, 198, 230])

```python
# Q5

sales = np.array([120, 135, 98,
                  160, 175, 190,
                  88, 92, 105,
                  210, 198, 230])

quarters = sales.reshape(4, 3)

print("Quarterly sales:")
print(quarters)

averages = quarters.mean(axis=1)

print("\nQuarter averages:")
print(averages)

highest_quarter = averages.argmax()

print("\nHighest average quarter index:", highest_quarter)
print("Highest average:", averages[highest_quarter])
```

Q6.
Build a 3x3 multiplication table from the array below using broadcasting only. Do not use loops and do not use np.outer — reshape one of the operands so that broadcasting produces the grid.

a = np.array([2, 3, 4])

```python
# Q6

a = np.array([2, 3, 4])

table = a[:, np.newaxis] * a[np.newaxis, :]

print(table)
print("Shape:", table.shape)
```
