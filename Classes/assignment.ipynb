```python
#Create a 6×6 identity matrix with integer dtype. The output must show no decimal points.

np.eye(6, dtype=int)
print(np.eye(6, dtype=int))
```

```python
#Create a 3×4 matrix where every element is 5, and report its shape, size, and ndim.

x = np.full((3, 4), 5)
print(x)
print("Shape:", x.shape)
print("Size:", x.size)
print("Number of dimensions:", x.ndim)
```

```python
#Given a = np.array([2, 4, 6]) and b = np.array([2.0, 4.0, 6.0]), check both arrays' dtype. Then convert b to integer dtype using astype(), and verify the dtype after conversion.


a= np.array([2, 4, 6])
b= np.array([2.0, 4.0, 6.0])
print("a dtype:", a.dtype)
print("b dtype:", b.dtype)
b_int = b.astype(int)
print("b converted to int dtype:", b_int.dtype)
```

```python
#Given arr = np.arange(10, 101, 10) - extract: (a) the 4th element, (b) the last element using negative indexing, (c) the 2nd-last element.

x = np.arange(10, 101, 10)
print("4th element:", x[3])
print("Last element:", x[-1])
print("2nd-last element:", x[-2])
```
