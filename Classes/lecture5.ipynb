```python
# Q5. Scholarship shortlisting
# Tasks

# Print the totals of students who scored above 250.
# Print the totals of students scoring above 180 but below 250.
# Print the totals of students who scored either below 150 or above 380.
# Print how many students fall in the band of task 2.
# Print the average of only those students who scored above 200.
# Print the index position of every student below 150.
# Print the highest total among the students who did not make the merit band of task 1.

import numpy as np
marks = np.array([[35, 40, 45, 50],
                  [69, 70, 65, 60],
                  [75, 80, 85, 70],
                  [44, 50, 55, 40],
                  [32, 30, 35, 30],
                  [90, 95, 100, 90]])
print(totals := np.sum(marks, axis=1))
print(totals[totals > 250])
print(totals[(totals > 180) & (totals < 250)])
print(totals[(totals < 150) | (totals > 380)])
print(np.sum((totals > 180) & (totals < 250)))
print(np.mean(totals[totals > 200]))
print(np.where(totals < 150)[0])
print(np.max(totals[totals <= 250]))
```

```python
# Q7. The result summary
# Tasks

# Print the class average and the middle value of the class.
# Print the gap between the highest and the lowest marks.
# Print the index positions of the topper and of the weakest student.
# Print the second highest marks in the class.
# Print how many students scored above the class average.
# Print how many students lie within one standard deviation of the class average.
# Print the class average after removing the topper.

import numpy as np
marks = np.array([[35, 40, 45, 50],
                  [69, 70, 65, 60],
                  [75, 80, 85, 70],    
                  [44, 50, 55, 40],
                  [32, 30, 35, 30],
                  [90, 95, 100, 90]])
totals = np.sum(marks, axis=1)
class_average = np.mean(totals)
middle_value = np.median(totals)
gap = np.max(totals) - np.min(totals)
topper_index = np.argmax(totals)
weakest_index = np.argmin(totals)
second_highest = np.partition(totals, -2)[-2]
above_average_count = np.sum(totals > class_average)
std_dev = np.std(totals)
within_one_std_dev_count = np.sum((totals >= (class_average - std_dev)) & (totals <= (class_average + std_dev)))
average_without_topper = np.mean(np.delete(totals, topper_index))
print("Class Average:", class_average)
print("Middle Value:", middle_value)
print("Gap between highest and lowest marks:", gap)
print("Index of Topper:", topper_index)
print("Index of Weakest Student:", weakest_index)
print("Second Highest Marks:", second_highest)
print("Number of students above class average:", above_average_count)
print("Number of students within one standard deviation of class average:", within_one_std_dev_count)
print("Class Average after removing topper:", average_without_topper)   
```

```python
# Q8. Subject-wise and student-wise reports
# Tasks

# Print the average of each subject and the total of each student.
# Print the index of the toughest subject for the class.
# Print the index of the student who topped overall.
# For every subject, print the marks of the student who topped that subject.
# Print how many students scored above 60 in each subject.
# Print the index of the subject in which the class marks are the most spread out.
# Print the total marks of the student who scored the lowest in Chemistry.

import numpy as np
marks = np.array([[35, 40, 45, 50],
                  [69, 70, 65, 60],
                  [75, 80, 85, 70],
                    [44, 50, 55, 40],
                    [32, 30, 35, 30],
                    [90, 95, 100, 90]])
subject_averages = np.mean(marks, axis=0)
toughest_subject_index = np.argmin(subject_averages)
student_totals = np.sum(marks, axis=1)
topper_index = np.argmax(student_totals)
topper_per_subject = np.max(marks, axis=0)
above_60_counts = np.sum(marks > 60, axis=0)
subject_spread = np.ptp(marks, axis=0)
most_spread_subject_index = np.argmax(subject_spread)
lowest_chemistry_index = np.argmin(marks[:, 2])
lowest_chemistry_total = student_totals[lowest_chemistry_index]
print("Subject Averages:", subject_averages)
print("Toughest Subject Index:", toughest_subject_index)
print("Topper Index:", topper_index)
print("Topper Marks per Subject:", topper_per_subject)
print("Number of Students above 60 in each subject:", above_60_counts)
print("Subject with most spread out marks Index:", most_spread_subject_index)
print("Total Marks of Student with Lowest Chemistry Score:", lowest_chemistry_total)    

```

```python
# Q10. Grace marks and weightage
# Tasks

# Apply the subject-wise grace marks to the whole class, keeping every value at or below 100.
# Multiply each subject by its own weightage and print the table rounded to 1 decimal.
# Produce a table showing how far each mark sits from that subject's own class average.
# Give 5 bonus marks to the first student, 0 to the second, 2 to the third and 0 to the fourth — in a single operation.
# From the table of task 3, print how many entries lie below the subject average.
# Produce a table where every subject is brought to one scale by subtracting its average and dividing by its spread. Round to 2 decimals.

import numpy as np
marks = np.array([[35, 40, 45, 50],
                  [69, 70, 65, 60],
                  [75, 80, 85, 70],
                  [44, 50, 55, 40],
                  [32, 30, 35, 30],
                  [90, 95, 100, 90]])
grace_marks = np.array([5, 10, 15, 20])
marks_with_grace = np.clip(marks + grace_marks, 0, 100)
weightage = np.array([0.2, 0.3, 0.4, 0.1])
weighted_marks = np.round(marks_with_grace * weightage, 1)
subject_averages = np.mean(marks_with_grace, axis=0)
deviation_from_average = marks_with_grace - subject_averages
bonus_marks = np.array([5, 0, 2, 0])    
marks_with_bonus = marks_with_grace + bonus_marks
total_below_average = np.sum(marks_with_bonus < subject_averages)
spread = np.ptp(marks_with_grace, axis=0)
scaled_marks = np.round((marks_with_grace - subject_averages) / spread, 2)
print("Marks with Grace:\n", marks_with_grace)
print("Weighted Marks:\n", weighted_marks)
print("Deviation from Subject Averages:\n", deviation_from_average)
print("Marks with Bonus:\n", marks_with_bonus)
print("Number of Entries Below Subject Average:", total_below_average)
print("Scaled Marks:\n", scaled_marks)
print("Subject Averages:", subject_averages)
```
