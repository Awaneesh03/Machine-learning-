```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
df = pd.read_csv("../Data/Titanic-Dataset.csv")
mpg = pd.read_csv("../Data/mpg.csv")
```

```python
print(df.shape,mpg.shape)
```

```python
df.head()
```

```python
plt.figure(figsize=(10,6))
sns.histplot(data=df,x='Fare',y='Age')
plt.title('Fare vs Age in Titanic Dataset')
plt.grid()
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.histplot(data=mpg,x ='displacement')
plt.title("Distribution of Displacement")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.boxplot(data=mpg,x='origin', y='mpg')
plt.title("MPG by Origin")
plt.show()
```
