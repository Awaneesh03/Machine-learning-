```python
#the company report calls first class "somewhat safer". find the survival rate of each class and say in one line wheather "somewhat" is honest
df.groupby("Pclass")["Survived"].mean().round(2)
```

```python
df.groupby("Pclass")["Survived"].mean().round(2)*100
```

**Answer:** No - "somewhat" understates it. First class survived at 63% vs 47% for second and 24% for third, so a first-class passenger was about 2.6x as likely to survive as a third-class one.
